```python
#reversing a list or array using recursion
def rev_func(num, left, right):
  if left>right or left==right:
    return num
  num[left],num[right]=num[right],num[left]
  return rev_func(num,left+1,right-1)

list1=[5,7,3,2,6,1,5,9]
var=rev_func(list1,2,5)
print(var)
```
