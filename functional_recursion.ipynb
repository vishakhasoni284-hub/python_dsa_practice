```python
#sum of n to 1 using functional recursion

def func(n):
  if n==1:
    return 1
  return n + func(n-1)
var=func(4)
print(var)
```

```python
def func(n):
  if n==1:
    return 1
  return n * func(n-1)
var=func(5)
print(var)
```
