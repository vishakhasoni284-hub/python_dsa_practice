```python
#print 1 to n (5) using head recursion
def func(i,n):
  if i>n:
    return
  print(i)
  func(i+1,n)

func(1,5)
```

```python
#print 1 to n by using tail recursion (backtracking)
def func(n):
  if n==0:
    return
  func(n-1)
  print(n)
func(5)
```

```python
#print n to 1 by using head recursion

def func(n):
  if n==0:
   return
  print(n)
  func(n-1)
func(5)
```

```python
#print n to 1 by using tail recursion (backtracking)
def func(i,n):
  if i>n:
    return
  func(i+1,n)
  print(i)
func(1,5)
```
