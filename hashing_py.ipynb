```python
n=[5,3,2,2,1,5,5,7,5,10]
m=[10,111,1,9,5,67,2]
hash_list=[0]*11
for num in n:
  hash_list[num]+=1
for j in range (0,len(m)):
  if m[j]<1 or m[j]>10:
    print(m[j] , " : " ,0)
  else:
    print(m[j]," : ",hash_list[m[j]])
```

```python
n=[5,3,2,2,1,5,5,7,5,10]
m=[10,111,1,9,5,67,2]
hash_dict={}
for num in n:
  if num not in hash_dict:
    hash_dict[num]=0
  else:
    hash_dict[num]+=1
for j in range (0,len(m)):
  if m[j]<1 or m[j]>10:
    print(m[j] , " : " ,0)
  else:
    print(m[j]," : ",hash_dict.get(m[j],0))
```
