```python
#find value of n of fibonacci addition output using recursion
def fibo(num,c,a,b,count):
  c=a+b
  if c==num :
    return count
  return fibo(num,c,a=b,b=c,count=count+1)

var=fibo(5,0,0,1,2)
print(var)




```

```python
#find fibonacci output of given number

def fibo(num):
  if num==0 or num==1 :
    return num
  return fibo(num-1)+fibo(num-2)
vaar=fibo(5)
print(vaar)
```
