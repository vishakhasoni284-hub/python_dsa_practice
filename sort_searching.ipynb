```python
from ast import Index
def linear_search(arr,target):
  size=len(arr)
  for i in range(0,size):
    if(arr[i]==target):
      return i
  return -1
mylist=[1,2,3,70,5,6]

target=8

result=linear_search(mylist,target)
print(result)
```

```python
def binary_search(arr,target):
  start=0
  size=len(arr)
  end=size-1
  while(start<=end):

    mid=(start+end)//2
    if(target==arr[mid]):
      return mid
    elif(target<arr[mid]):
      end=mid-1
    elif(target>arr[mid]):
      start=mid+1

  return("not found")


myy_list=[10,23,35,45,50,70,85]
target=50
yes_no=binary_search(myy_list,target)
print(yes_no)



```

```python
def bubble_sort(arr):
  end=len(arr)-1

  for i in range(0,end):
    start=0
    next=start+1
    while(next<=end):
      if (arr[start]>arr[next]):
        arr[start],arr[next]=arr[next],arr[start]
      start=start+1
      next=next+1
  return arr


new_list=[64, 34, 25, 12, 22, 11, 90]
var=bubble_sort(new_list)
print(var)

```

```python
def selection_sort(arr):
  for i in range (0,len(arr)-1):
    temp=0
    min=i
    z=i+1
    while(z<=len(arr)-1):
      if arr[z]<arr[min]:
        min=z
      z=z+1
    arr[i],arr[min]=arr[min],arr[i]

  return arr

list1=[64, 25, 12, 22, 11]
vaar=selection_sort(list1)
print(vaar)
```

```python
def insertion_sort(arr):
  for i in range(1,len(arr)):
    temp=arr[i]
    j=i-1
    while(j>=0):
      if(temp>arr[j]):
        break
      else:
        arr[j+1]=arr[j]#shift element to +1 position
      j=j-1 #for example if we compared with position j=2 it will become 1
      arr[j+1]=temp #as we have alredy made swap of j=2 with j=3 j=2 is empty
      #and j has become 1 (j=j-1) so we will put the temp or key value there
  return arr

list1=[64, 25, 12, 22, 11]
vaar=insertion_sort(list1)
print(vaar)





```
