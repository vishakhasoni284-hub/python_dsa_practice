```python
# check palindrom
#we reveser the thing either its string or number
# for number

def palindrom(n):
  result=0
  n1=n
  while (n>0):
    num=n%10
    n=n//10
    result=result*10+num
  if result==n1:
    return ('Yes a palindrom.')
  else:
    return('Not a palindrom.')
print("Enter a number to check palindrome.")
var=int(input())
var1=palindrom(var)
print(var1)
```

```python
#armstrong number
def armstrong(n):
  num=n
  count=0
  while(num>0):
    num=num//10
    count=count+1

  add=0
  num=n
  while(num>0):
    var=num%10
    num=num//10
    add=add+var**count

  if (add==n):
    return ("Yes an armstrong number.")
  else:
    return ("Not an armstrong number.")

print("Enter a number to check armstrong number.")
var=int(input())
var1=armstrong(var)
print(var1)
```

```python
def armstrong(n):
  num=n
  power=len(str(n))

  add=0
  num=n
  while(num>0):
    var=num%10
    num=num//10
    add=add+var**power

  if (add==n):
    return ("Yes an armstrong number.")
  else:
    return ("Not an armstrong number.")

print("Enter a number to check armstrong number.")
var=int(input())
var1=armstrong(var)
print(var1)
```

```python
def factors(n):
  num=1
  list1=[]
  while(num<=n//2):
    if num==1:
      list1.append(num)
    elif n%num==0:
      list1.append(num)
    num=num+1
  list1.append(n)
  return list1

print("Enter a number to find its factors.")
var=int(input())
var1=factors(var)
print(var1)

```

```python
# another way to do that
from math import sqrt
print("Enter a number to find its factors.")
num=int(input())
result=[]
for i in range (1,int(sqrt(num))+1):
  if num%i==0:
    result.append(i)
  if num//i!=i:
    result.append(num//i)
result.sort()
print(result)
```
